```python
import pandas as pd
data = {
    'Name': ['Ravi', 'Sita', 'Ram', 'Priya', 'John'],
    'Age': [20, 21, 19, 22, 20],
    'Marks': [85, 90, 78, 95, 88],
    'City': ['Guntur', 'Vijayawada', 'Hyderabad', 'Chennai', 'Eluru']
}

df = pd.DataFrame(data, index=['A', 'B', 'C', 'D', 'E'])

print(df)
```

```python
print(df.loc[:, 'Name'])
```

```python
print(df.loc[:, 'Name'])
```

```python
print(df.loc[['A', 'B']])
```

```python
print(df.loc[['A', 'B'], ['Name', 'Marks']])
```

```python
print(df.loc[df['Marks'] > 85])
```

```python
print(df.iloc[0])
```

```python
print(df.iloc[:, 0])
```

```python
print(df.iloc[0:2])
```

```python
print(df.iloc[[0, 1], [0, 2]])
```

```python
print(df.iloc[0:3, 0:2])
```
